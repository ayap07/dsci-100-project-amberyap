#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
#seed has already been set - from previous - student #
clean_model <- clean_data |> #pipelines clean_data info into clean_model
    mutate( #Tourney wins is originally a factor so numerify it 
            won_first_game = if_else(as.numeric(as.character(tourney_wins)) >= 1, 1, 0) 
        #tourney_wins where 1 = won first game, 0 = lost first game
        ) |> 
    select(#Selects the predictors to use in the model 
        won_first_game, 
        seed, 
        reg_wins_pct, 
        conf_wins_pct, 
        conf_rank
        ) |> 
    drop_na() #drop rows with missing values
head(clean_model) #double checking by pulling up the first 6
```

```r
clean_split <- initial_split(clean_model, prop = 0.8, strata = won_first_game)
train_data <- training(clean_split) #splits the data 80/20 for training/testing 
test_data <- testing(clean_split) 
lm_recipe <- recipe( #sets the recipe to predict first game win from predictors
    won_first_game ~ seed + reg_wins_pct + conf_wins_pct + conf_rank, 
    data = train_data #using training data set to train the model
    )
lm_spec <- linear_reg() |> #making the model specification for linear regression model
    set_engine("lm") |> 
    set_mode("regression") 
lm_workflow <- workflow() |> #workflow to fit the data with the model and recipe
    add_recipe(lm_recipe) |> 
    add_model(lm_spec)
cv_folds <- vfold_cv(train_data, v = 5, strata = won_first_game) #make 5 fold cross 
cv_results <- lm_workflow |> #using the lm_workflow to then fit resamples of it
    fit_resamples(resamples = cv_folds) #checking resamples 
collect_metrics(cv_results) #collecting the results of it by the rmse and rsq - rmse is only relevant one
```

```r
#add in the 5-fold validation cross and then run with test data to see the confusion matrix after
lm_fit <- lm_workflow |> #fitting workflow to training data
    fit(data = train_data) 
lm_preds <- lm_fit |> #uses the fit data to predict for test data
    predict(test_data) |> #predicting by using test data that we saved aside until now 
    bind_cols(test_data) #binding the columns 
lm_test_results <- lm_preds |> #now making the results for rmse 
    metrics(truth = won_first_game, estimate = .pred) |> 
    filter(.metric == "rmse") #grabbing the rmse value from after predicting model
lm_test_results #displaying the rmse value 
```

```r
#Now is to determine the accuracy of model via conf_matrix 
test_classified <- lm_preds |> #taking the test set prediction of team's 
#continuous linear regression and converting into binary classification
    mutate(pred_class = if_else(.pred >0.5, 1, 0))
#if .pred >0.5, =1, 0 if it isn't
head(test_classified)#showing the first 6 rows of data set 

conf_matrix <- table( #making a 2x2 matrix to see actual vs. predicted
    #wins and determine whether true loss, false win, false loss, or true win
    Actual = test_classified$won_first_game, #going into classified data to grab first game
    Predicted = test_classified$pred_class #grabbing the predicition class 
    )
conf_matrix #displaying the 2x2 table 
```
